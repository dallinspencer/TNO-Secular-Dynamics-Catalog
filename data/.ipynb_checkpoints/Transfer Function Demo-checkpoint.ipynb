```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def get_range(df, arange, erange, Irange, osc_or_prop=True):
    amid = np.mean(arange); adist = abs(arange[0] - arange[1])/2
    emid = np.mean(erange); edist = abs(erange[0] - erange[1])/2
    Imid = np.mean(Irange); Idist = abs(Irange[0] - Irange[1])/2
    
    if osc_or_prop:
        inda = np.where(abs(df['OscSMA'] - amid) <= adist)[0]
        inde = np.where(abs(df['OscEcc'] - emid) <= edist)[0]
        indI = np.where(abs(df['OscInc']*180/np.pi - Imid) <= Idist)[0]
    else:
        inda = np.where(abs(df['PropSMA'] - amid) <= adist)[0]
        inde = np.where(abs(df['PropEcc'] - emid) <= edist)[0]
        indI = np.where(abs(np.arcsin(df['PropSinI'])*180/np.pi - Imid) <= Idist)[0]

    indget = np.intersect1d(inda, inde)
    indget = np.intersect1d(indget, indI)

    return df.iloc[indget]


grid_df = pd.read_csv('ctno_grid_pe_csv.zip', index_col=0)
tno_df = pd.read_csv('AllTNOs_500myr_pe.csv', index_col=0)
print(grid_df.head())
print(tno_df.head())
```

```python
sub_grid_df = get_range(grid_df, (42.3, 44.5), (0,0.2), (0,10), osc_or_prop=True)
sub_tno_df = get_range(tno_df, (42.3, 44.5), (0,0.2), (0,10), osc_or_prop=True)

proper_elements = np.column_stack([np.sqrt(5/4)*grid_df['PropSMA']/np.mean(grid_df['PropSMA']), 
                                   np.sqrt(2)*grid_df['PropEcc'], 
                                   np.sqrt(2)*grid_df['PropSinI']])

from scipy.spatial import KDTree
tree = KDTree(proper_elements)

def hcm(a, rmsa, rmse, rmsi):   
    G = 6.673e-11
    M = 1.989e30
    n = np.sqrt(G*M/(a*1.496e11)**3)
    return n*a*1.498e11*np.sqrt(5/4*(rmsa/a)**2+2*rmse**2+2*rmsi**2)


def get_transfer_weights(a_obs, e_obs, sini_obs, hcm_obs, k=200, a0=None, e0=None, i0=None, tree = None):
    """
    Returns initial conditions of k nearest neighbors in PROPER
    element space, with weights based ONLY on proper element distance.
    Initial conditions are carried along passively.
    Input i0 is in units of radians
    Input ip_obs is in sinI units. 

    ap, ep, ip are proper elements of your sample
    i0, a0, e0 is osculating (t=0) elements of your 
    """
    query = np.array([[np.sqrt(5/4)*a_obs/40, np.sqrt(2)*e_obs, np.sqrt(2)*sini_obs]])
    dists, idx = tree.query(query, k=k)

    h = np.median(dists)   # adaptive bandwidth
    stab = np
    weights = np.exp(-0.5 * (dists[0] / h)**2) 
    weights /= weights.sum()

    epsilon = np.percentile(hcm_obs, 5)  # floor at 5th percentile
    stab_weights = 1.0 / (hcm_obs[idx[0]] + epsilon)

    # Normalize stability weights independently before combining
    stab_weights /= stab_weights.sum()

    # Combined weight
    combined = weights * stab_weights
    combined /= combined.sum()

    # Initial conditions are just passengers — not used for weighting
    return a0[idx[0]], e0[idx[0]], i0[idx[0]], weights, idx[0]


print("Pre-computing transfer weights for all observed objects...")

transfer_i0 = []   # list of arrays, one per observed object
transfer_w  = []   # corresponding weights
transfer_a0 = []
transfer_e0 = []
transfer_inds = []

#cluster = hcm(sub_tno_df['PropSMA'], sub_tno_df['PropSMA_err'], sub_tno_df['PropEcc_err'], sub_tno_df['PropSinI_err'])
cluster = hcm(grid_df['PropSMA'], grid_df['PropSMA_err'], grid_df['PropEcc_err'], grid_df['PropSinI_err'])

for ap_k, ep_k, ip_k in zip(tno_df['PropSMA'], tno_df['PropEcc'], tno_df['PropSinI']):
    a0_nn, e0_nn, i0_nn, w_nn, inds = get_transfer_weights(ap_k, ep_k, ip_k, cluster.to_numpy(), k=5, tree=tree, 
                                                           a0=grid_df['OscSMA'].to_numpy(),
                                                           e0=grid_df['OscEcc'].to_numpy(), 
                                                           i0=grid_df['OscInc'].to_numpy())
    #i0_nn, a0_nn, e0_nn, w_nn, inds = get_transfer_weights(ap_k, ip_k, ep_k, cluster, k=20, tree=tree1, a0=np.array(A2), i0=np.array(I2), e0=np.array(E2))
    transfer_i0.append(i0_nn)
    transfer_a0.append(a0_nn)
    transfer_e0.append(e0_nn)
    transfer_w.append(w_nn)
    transfer_inds.append(inds)

# Convert to arrays if k is fixed (faster indexing)
transfer_i0 = np.array(transfer_i0)  # shape (n_obs, k)
transfer_w  = np.array(transfer_w)   # shape (n_obs, k)
transfer_a0 = np.array(transfer_a0)  # shape (n_obs, k)
transfer_e0 = np.array(transfer_e0)  # shape (n_obs, k)
transfer_inds  = np.array(transfer_inds)   # shape (n_obs, k)

```

```python
bina = np.linspace(42.3, 44.5, 23)
bine = np.linspace(0,0.3,31)
binI = np.linspace(0,10,21)

bina2 = np.linspace(42.3, 44.5, 22)
bine2 = np.linspace(0,0.3,30)
binI2 = np.linspace(0,10,20)

histe, edge1, edge2 = np.histogram2d(transfer_a0.flatten(), transfer_e0.flatten(), bins=(bina, bine), weights=transfer_w.flatten())
histI, edge1, edge2 = np.histogram2d(transfer_a0.flatten(), transfer_i0.flatten(), bins=(bina, binI), weights=transfer_w.flatten())

from matplotlib.colors import LogNorm
fig,ax = plt.subplots(1,2,figsize=(8,3))
c1 = ax[0].contourf(bina2, bine2, histe.T + 1e-2, levels=np.geomspace(1e-1,2e1, 28), norm=LogNorm(), extend='both', cmap='Blues')

ax[0].scatter(sub_tno_df['PropSMA'], sub_tno_df['PropEcc'], s=4, alpha=0.6, c='tab:red', label='TNO Proper Elements')

ax[0].set_xlim(42.3, 44.5)
ax[0].set_ylim(0,0.3)
ax[0].set_xlabel('SMA (AU)', fontsize=12)
ax[0].set_ylabel(r'Ecc', fontsize=12)
ax[0].set_title('Classical Belt Transfer Function', fontsize=14)


c1 = ax[1].contourf(bina2, binI2, histI.T + 1e-2, levels=np.geomspace(1e-1,2e1, 28), norm=LogNorm(), extend='both', cmap='Blues')
ax[1].scatter(sub_tno_df['PropSMA'], 180/np.pi*np.arcsin(sub_tno_df['PropSinI']), s=4, alpha=0.6, c='tab:red', label='TNO Proper Elements')

ax[1].set_xlim(42.3, 44.5)
ax[1].set_ylim(0,10)
ax[1].set_xlabel('SMA (AU)', fontsize=12)
ax[1].set_ylabel(r'Inc $(^{\circ})$', fontsize=12)
ax[1].set_title('Classical Belt Transfer Function', fontsize=14)

plt.show()
```
